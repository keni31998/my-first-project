```python
# ==============================================
# 集团营销中心与子公司经营分析交互平台
# 10大维度 | 8大图表 | 全交互 | WRDS真实数据
# ==============================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings("ignore")

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

# ==============================================
# 🔥 交互 1：输入集团 + 子公司代码
# ==============================================
print("=" * 60)
print("    集团 - 子公司 经营分析系统")
print("=" * 60)

tickers = input("输入 集团、子公司代码（逗号分隔）：\n例如：AAPL,MSFT,TSLA\n").strip().upper().split(",")
start_year = int(input("开始年份："))
end_year = int(input("结束年份："))

tickers = [t.strip() for t in tickers]
print(f"\n正在加载：{tickers} | 年份：{start_year}-{end_year}")

# ==============================================
# 从 WRDS 提取全部数据
# ==============================================
dfs = []
for tic in tickers:
    sql = f"""
        SELECT
            '{tic}' AS company,
            fyear,
            sale     AS revenue,    -- 营收
            ni       AS profit,     -- 净利润
            at       AS assets,     -- 总资产
            (ni/sale) AS profit_margin, -- 净利率
            (ni/at)   AS roe           -- ROE
        FROM comp.funda
        WHERE tic = '{tic}'
          AND fyear BETWEEN {start_year} AND {end_year}
        ORDER BY fyear
    """
    df = db.raw_sql(sql)
    dfs.append(df)

df = pd.concat(dfs).dropna()
df["fyear"] = df["fyear"].astype(int)

# 计算增长率
df = df.sort_values(["company", "fyear"])
df["revenue_growth"] = df.groupby("company")["revenue"].pct_change() * 100
df["profit_growth"] = df.groupby("company")["profit"].pct_change() * 100

print("\n✅ 数据加载完成！")
print(df.round(2))

# ==============================================
# 计算集团总和 & 子公司占比
# ==============================================
def get_group_summary(df):
    group = df.groupby("fyear").agg({
        "revenue": "sum",
        "profit": "sum",
        "assets": "sum"
    }).rename(columns={
        "revenue": "group_revenue",
        "profit": "group_profit",
        "assets": "group_assets"
    })
    return group

group_df = get_group_summary(df)
df = df.merge(group_df, on="fyear")
df["revenue_share"] = df["revenue"] / df["group_revenue"] * 100
df["profit_share"] = df["profit"] / df["group_profit"] * 100

latest = df[df["fyear"] == df["fyear"].max()]

# ==============================================
# 🔥 交互菜单：8大功能模块
# ==============================================
while True:
    print("\n" + "="*50)
    print("📊 功能菜单（输入数字选择）")
    print("="*50)
    print("1  → 营收对比（柱状图）")
    print("2  → 净利润对比（折线图）")
    print("3  → 子公司营收占比（饼图）")
    print("4  → 子公司利润贡献（条形图）")
    print("5  → 营收/利润增长率（双柱状图）")
    print("6  → 盈利能力排名（ROE/净利率）")
    print("7  → 总资产规模对比（面积图）")
    print("8  → 综合评分雷达图")
    print("0  → 退出")
    
    choice = input("\n请选择功能：")

    # -------------------------------------------------
    # 模块1：营收对比
    # -------------------------------------------------
    if choice == "1":
        plt.figure(figsize=(12,5))
        for c in df["company"].unique():
            sub = df[df["company"]==c]
            plt.bar(sub["fyear"].astype(str)+c, sub["revenue"], label=c)
        plt.title("集团 vs 子公司 营收对比")
        plt.ylabel("营收")
        plt.legend()
        plt.xticks(rotation=30)
        plt.grid(alpha=0.3)
        plt.show()

    # -------------------------------------------------
    # 模块2：净利润对比
    # -------------------------------------------------
    elif choice == "2":
        plt.figure(figsize=(12,5))
        for c in df["company"].unique():
            sub = df[df["company"]==c]
            plt.plot(sub["fyear"].astype(str), sub["profit"], marker="o", label=c)
        plt.title("集团 vs 子公司 净利润趋势")
        plt.ylabel("净利润")
        plt.legend()
        plt.grid()
        plt.show()

    # -------------------------------------------------
    # 模块3：子公司营收占比（饼图）
    # -------------------------------------------------
    elif choice == "3":
        plt.figure(figsize=(8,8))
        plt.pie(latest["revenue_share"], labels=latest["company"], autopct="%.1f%%")
        plt.title("子公司营收占集团比重")
        plt.show()

    # -------------------------------------------------
    # 模块4：利润贡献度（条形图）
    # -------------------------------------------------
    elif choice == "4":
        plt.figure(figsize=(10,5))
        plt.barh(latest["company"], latest["profit_share"])
        plt.title("子公司利润贡献度")
        plt.xlabel("占比 %")
        plt.grid(alpha=0.3)
        plt.show()

    # -------------------------------------------------
    # 模块5：增长率对比
    # -------------------------------------------------
    elif choice == "5":
        plt.figure(figsize=(12,5))
        width = 0.35
        years = latest["fyear"].astype(str).values
        
        for i, c in enumerate(df["company"].unique()):
            sub = df[df["company"]==c]
            x = np.arange(len(sub))
            plt.bar(x + i*width, sub["revenue_growth"].fillna(0), width, label=f"{c} 营收增速")
        
        plt.title("营收增长率对比")
        plt.xticks(x, sub["fyear"].astype(str))
        plt.legend()
        plt.grid(alpha=0.3)
        plt.show()

    # -------------------------------------------------
    # 模块6：盈利能力排名
    # -------------------------------------------------
    elif choice == "6":
        plt.figure(figsize=(10,5))
        plt.barh(latest["company"], latest["profit_margin"]*100)
        plt.title("净利率排名")
        plt.xlabel("%")
        plt.show()

    # -------------------------------------------------
    # 模块7：资产规模对比（面积图）
    # -------------------------------------------------
    elif choice == "7":
        plt.figure(figsize=(12,5))
        for c in df["company"].unique():
            sub = df[df["company"]==c]
            plt.fill_between(sub["fyear"].astype(str), sub["assets"], alpha=0.5, label=c)
        plt.title("资产规模对比")
        plt.legend()
        plt.grid()
        plt.show()

    # -------------------------------------------------
    # 模块8：综合评分雷达图
    # -------------------------------------------------
    elif choice == "8":
        from math import pi
        cats = ["营收规模", "利润", "净利率", "增速", "资产"]
        N = len(cats)

        for c in latest["company"].unique():
            row = latest[latest["company"]==c].iloc[0]
            values = [
                row["revenue"] / latest["revenue"].max() * 10,
                row["profit"] / latest["profit"].max() * 10,
                row["profit_margin"] * 10,
                max(row["revenue_growth"],0) / 50 *10 if not pd.isna(row["revenue_growth"]) else 0,
                row["assets"] / latest["assets"].max() *10
            ]
            values += values[:1]
            angles = [n / float(N) * 2 * pi for n in range(N)]
            angles += angles[:1]

            plt.figure(figsize=(6,6))
            ax = plt.subplot(111, polar=True)
            ax.plot(angles, values)
            ax.fill(angles, values, alpha=0.3)
            plt.xticks(angles[:-1], cats)
            plt.title(f"{c} 综合评分")
            plt.show()

    elif choice == "0":
        print("👋 退出成功")
        break
    
    else:
        print("输入错误，请重试")
```

```python
# ==============================
# 集团-子公司经营分析平台 Streamlit 完整版
# 包含：WRDS登录 + 10大维度 + 8张图表
# ==============================
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import wrds
from math import pi
import warnings
warnings.filterwarnings("ignore")

# 页面设置
st.set_page_config(layout="wide")
st.title("📊 集团营销中心与子公司经营分析系统")

# ======================
# 1) WRDS 登录（必须写在这里）
# ======================
st.sidebar.header("🔐 WRDS 登录")
username = st.sidebar.text_input("用户名")
password = st.sidebar.text_input("密码", type="password")

if st.sidebar.button("登录"):
    try:
        db = wrds.Connection(wrds_username=username, wrds_password=password)
        st.session_state["db"] = db
        st.sidebar.success("✅ 登录成功")
    except:
        st.sidebar.error("❌ 登录失败")

# ======================
# 2) 输入公司与年份
# ======================
st.sidebar.header("📌 分析设置")
tickers = st.sidebar.text_input("公司代码（逗号分隔）", "AAPL,MSFT,TSLA")
start_year = st.sidebar.number_input("开始年份", 2018, 2023, 2018)
end_year = st.sidebar.number_input("结束年份", 2018, 2023, 2023)

# ======================
# 3) 读取 WRDS 数据
# ======================
if st.sidebar.button("🚀 开始分析"):
    if "db" not in st.session_state:
        st.warning("请先登录！")
        st.stop()

    db = st.session_state["db"]
    ticker_list = [x.strip().upper() for x in tickers.split(",")]
    
    dfs = []
    for tic in ticker_list:
        sql = f"""
            SELECT
                '{tic}' AS company,
                fyear,
                sale AS revenue,
                ni AS profit,
                at AS assets,
                (ni/sale) AS profit_margin,
                (ni/at) AS roe
            FROM comp.funda
            WHERE tic = '{tic}'
            AND fyear BETWEEN {start_year} AND {end_year}
        """
        df = db.raw_sql(sql)
        dfs.append(df)

    df = pd.concat(dfs).dropna()
    df["fyear"] = df["fyear"].astype(int)
    df = df.sort_values(["company", "fyear"])
    df["revenue_growth"] = df.groupby("company")["revenue"].pct_change() * 100
    df["profit_growth"] = df.groupby("company")["profit"].pct_change() * 100

    # 计算占比
    group = df.groupby("fyear").agg({"revenue":"sum","profit":"sum"}).rename(columns={"revenue":"g_rev","profit":"g_prof"})
    df = df.merge(group,on="fyear")
    df["rev_share"] = df["revenue"]/df["g_rev"]*100
    df["prof_share"] = df["profit"]/df["g_prof"]*100

    st.session_state["df"] = df
    st.success("✅ 数据加载完成")
    st.dataframe(df.round(2))

# ======================
# 4) 图表模块
# ======================
if "df" in st.session_state:
    df = st.session_state["df"]
    latest = df[df["fyear"] == df["fyear"].max()]

    st.subheader("1️⃣ 营收对比（柱状图）")
    fig, ax = plt.subplots()
    for c in df["company"].unique():
        s = df[df["company"]==c]
        ax.bar(s["fyear"].astype(str), s["revenue"], label=c)
    ax.legend()
    st.pyplot(fig)

    st.subheader("2️⃣ 净利润趋势（折线图）")
    fig, ax = plt.subplots()
    for c in df["company"].unique():
        s = df[df["company"]==c]
        ax.plot(s["fyear"].astype(str), s["profit"], marker="o", label=c)
    ax.legend()
    ax.grid(True)
    st.pyplot(fig)

    st.subheader("3️⃣ 子公司营收占比（饼图）")
    fig, ax = plt.subplots()
    ax.pie(latest["rev_share"], labels=latest["company"], autopct="%.1f%%")
    st.pyplot(fig)

    st.subheader("4️⃣ 利润贡献度（条形图）")
    fig, ax = plt.subplots()
    ax.barh(latest["company"], latest["prof_share"])
    st.pyplot(fig)
```
